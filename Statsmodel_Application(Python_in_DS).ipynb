```python
#Importing the necessary libraries
%pip install pandas numpy matplotlib scipy statsmodels scikit-learn
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
import statsmodels.formula.api as smf
```

```python
#Generate a fictional dataset of 120 students
np.random.seed(42)
n = 120
study_hours = np.random.uniform(1,15,n)
attendance = np.random.uniform(50,100,n)
previous_score = np.random.uniform(35,90,n)
noise = np.random.normal(0,6,n)
exam_score = (10 + 1.8*study_hours + 0.25*attendance + 0.4*previous_score + noise)
df = pd.DataFrame({"Study Hours":study_hours, "Attendance":attendance, "Previous Score":previous_score, "Exam Score":exam_score})
df.head()
```

```python
#Inspect the Dataset
print(df.shape)
print(df.info())
print(df.isna().sum())
print(df.describe().round(2))
```

```python
#Using Linear Regression with Statsmodel
simple_model = smf.ols(formula="exam_score ~ study_hours", data=df).fit()
print(simple_model.summary())
```
